```python
import glob
import os.path as path

import numpy as np
from fake_spectra.plot_spectra import PlottingSpectra
from fake_spectra.spectra import Spectra
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d

from lyaemu import lyman_data
from lyaemu import flux_power
```

```python
# Flux power spectrum post analysis
def find_spectra_file(i, base="output"):
    # Find the saved spectra .hdf5 file in the snapshot's SPECTRA_NNN folder
    files = glob.glob(path.join(base, "SPECTRA_%03d" % i, "*.hdf5"))
    if len(files) != 1:
        raise IOError("Expected one spectra file for snapshot %d, found: %s" % (i, files))
    return path.basename(files[0])

def post_analysis(i, tau):
    # To just spectra
    gs = Spectra(i, "output", MPI=None, res=1.0, savefile=find_spectra_file(i), cofm=None, axis=None)
    fps = gs.get_flux_power_1D("H",1,mean_flux_desired=np.exp(-tau))     # get 1D flux power spectrum
    # Save spectra to file
    np.savetxt("output/SPECTRA_%03d/flux_power.txt" % i, fps, fmt="%.4e")
    return fps

# Do the chi2 analysis for the flux power spectrum
def chi_squared(expected, observed, sigma_sq):
    # observed is the fit, expected is the original spectrum
    return (expected - observed)**2/sigma_sq
```

```python
# Optimization for ngrid=200
# Three redshifts
scale_factors = [0.1908, 0.1792, 0.1681]
z_list = [1 / a - 1 for a in scale_factors]
obser_data = lyman_data.BoeraData()
kf = obser_data.get_kf()    # km / s

# Best fit results
z_tau = [1.0123737 ,1.24574982,1.52008839]
chi_2 = [9.52799037,69.26261864,35.06637876]

# Extract the observed power spectrum for different redshifts
# z = [5.0, 4.6, 4.2]
data_power  = obser_data.pf.reshape(-1, kf.shape[0]) #km / s; n_z * n_k # n_z * n_k
sigma_sq    = obser_data.covar_diag.reshape(-1, kf.shape[0])

# Auto-detect the available snapshots (e.g. [7, 8, 10]); the last snapshot
# maps to the lowest redshift, so iterate from the end of the list.
snaps = sorted(int(path.basename(d)[len("SPECTRA_"):])
               for d in glob.glob(path.join("output", "SPECTRA_[0-9][0-9][0-9]")))[-len(z_list):]
print("Using snapshots:", snaps)
```

```python
# Plot the column density function
snap_ind = snaps[-1]
ps = PlottingSpectra(num=snap_ind,base="output", savefile=find_spectra_file(snap_ind), label=f"num={snap_ind}")
# ps.plot_cddf("H",1)
ps.plot_spectrum("H",1,1215,9)
```

```python
# Same chi2 plot, but with a single shared y axis on the left
fig, axes = plt.subplots(1, len(z_list), figsize=(5 * len(z_list), 5),
                         sharex=True, sharey=True)
for i, z in enumerate(z_list):
    snap_ind = snaps[-1 - i]

    sim = np.loadtxt("output/SPECTRA_%03d/flux_power.txt" % (snap_ind))
    skf = sim[0,:]          # km / s
    sim_power = sim[1,:]    # P_F(k) for the simulated data at the current redshift

    # Rebinning simulated data to match the k bins of the observed data
    interp_func = interp1d(skf, sim_power, kind='linear', bounds_error=False, fill_value="extrapolate")
    sim_power_rebinned = interp_func(kf)

    chi2 = chi_squared(sim_power_rebinned, data_power[i], sigma_sq[i])

    # 1-sigma error bar for every data point, from the diagonal of the covariance matrix
    data_err = np.sqrt(sigma_sq[i])

    ax = axes[2-i]
    ax.loglog(kf, sim_power_rebinned, color="C1", label="Sim", zorder=1)
    # ls="none": each k bin is an isolated marker with its own vertical error bar
    ax.errorbar(kf, data_power[i], yerr=data_err, fmt="o", ls="none", color="C0",
                ms=3, elinewidth=1.5, capsize=3, capthick=1.5, zorder=2,
                label="Data z = %.1f" % z)
    ax.set_xscale("log")
    ax.set_yscale("log")
    ax.set_xlabel("k (s/km)", fontsize=12)
    ax.text(0.65, 0.82, f"Chi-squared: {np.sum(chi2):.1f}", transform=ax.transAxes)
    ax.text(0.65, 0.77, f"Desired $\\tau$: {z_tau[i]:.2f}", transform=ax.transAxes)
    ax.set_ylim(1.5e-1, 3e2)
    ax.legend()

# Only the leftmost panel keeps the y axis label and tick labels
axes[0].set_ylabel("P$_{F}$(k)", fontsize=12)
for ax in axes[1:]:
    ax.tick_params(axis="y", which="both", labelleft=False)

fig.subplots_adjust(wspace=0.05)
plt.show()
```

```python
# Compare the differences of ratio of P_b/P_cdm with the linear theory prediction
# P_b/P_cdm = T_b^2(k)/T_cdm^2(k)

# Read the transfer function results
import sys
sys.path.append("..")
from read_transfer import read_transfer

scale_factors = [0.1681, 0.1792, 0.1908]
redshifts = [1 / a - 1 for a in scale_factors]

trans = read_transfer("theory/ics_transfer_4.6.dat")
plt.plot(trans["k"], (trans["d_cdm"])**2)
plt.plot(trans["k"], (trans["d_b"])**2)
plt.ylim(1e5, 1e9)
plt.xlim(1e-3, 1e3)
plt.xscale('log')
plt.yscale('log')
```

```python
def load_power(filename):
    """Load a power spectrum file with columns k, power, modes.
    Removes rows where k is nan or where k/power are zero."""
    k, power, modes = np.loadtxt(filename, unpack=True)
    mask = np.isfinite(k) & np.isfinite(power) & (k > 0) & (power > 0)
    return k[mask], power[mask], modes[mask]

power_data = {}
for a in scale_factors:
    k_dm, power_dm, modes_dm = load_power(f"output/power-DM-{a}.txt")
    k_bar, power_bar, modes_bar = load_power(f"output/power-bar-{a}.txt")
    power_data[a] = {"dm": (k_dm, power_dm), "bar": (k_bar, power_bar)}
```

```python
# Read the matter power spectrum

fig, axes = plt.subplots(1, len(scale_factors), figsize=(5* len(scale_factors), 5), sharex=True)

for ax, a, z in zip(axes, scale_factors, redshifts):
    trans = np.loadtxt(f"theory/ics_matterpow_{z:.1f}.dat", )
    ax.plot(trans[:, 0], trans[:, 1], label="matter PS")
    # ax.plot(k_dm * 1e3, power_bar / power_dm)
    ax.set_xscale('log')
    ax.set_xlabel("k [h/Mpc]")
    ax.set_title(f"z = {z:.1f}")
    ax.set_xlim(1e-4, 1e0)
    ax.legend()

# axes[0].set_ylabel(r"P$_b$/P$_{DM}$")
# for ax in axes[1:]:
#     ax.tick_params(axis="y", which="both", labelleft=False)

fig.subplots_adjust(wspace=0.00)
plt.tight_layout()
```
